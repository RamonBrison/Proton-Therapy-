```python
import matplotlib.pyplot as plt
```

### CSDA Data from PSTAR

```python
water_CSDA_data = {"7.000E+01": "4.080E+00",
                       "1.000E+02": '7.718E+00',
                       '1.200E+02': '1.066E+01',
                       '1.500E+02': '1.577E+01',
                       '1.750E+02': '2.062E+01',
                       '2.000E+02': '2.596E+01',
                       '2.250E+02': '3.174E+01',
                       '2.500E+02': '3.794E+01'}
```

### Range Calculation

```python
water_distances = []

for energy, csda in water_CSDA_data.items():
    distance = float(csda) * 0.998
    water_distances.append(distance)

print(water_distances)
```

## Plots

```python
energy = [float(e) for e in water_CSDA_data.keys()]
stopping_power = [float(s) for s in water_CSDA_data.values()]

fig, (ax1, ax2) = plt.subplots(1,2, figsize=(12,5))

ax1.plot(energy, stopping_power, marker='o', color='royalblue', linestyle='none')
ax1.set_xlabel('Energy (MeV)')
ax1.set_ylabel('Stopping Power (MeV·cm²/g)')
ax1.set_title('Energy x Stopping Power')
ax1.grid(True, alpha=0.6)

ax2.plot(energy, water_distances, marker='o', color='tomato', linestyle='none')
ax2.set_xlabel('Energy (MeV)')
ax2.set_ylabel('Range (cm)')
ax2.set_title('Energy x Range')
ax2.grid(True, alpha=0.6)
```
